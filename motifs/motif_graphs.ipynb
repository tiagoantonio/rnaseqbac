```python
import pandas as pd
import matplotlib.pyplot as plt
import glob
import os

# 1. Load motif results
motif_df = pd.read_csv("motif_hits.csv")
motif_genes = set(motif_df["gene_id"].unique())

# 2. Loop over all DE result files in a folder (adjust pattern as needed)
de_files = glob.glob("de_results/*.csv")  # assumes all DE files are here

for de_file in de_files:
    # 3. Load DE result
    df = pd.read_csv(de_file)
    
    # Assume first column is gene name (some tools like DESeq2 export it that way)
    if df.columns[0] == "":
        df.rename(columns={df.columns[0]: "gene_id"}, inplace=True)
    else:
        df.rename(columns={df.columns[0]: "gene_id"}, inplace=True)

    # 4. Add a column to indicate motif presence
    df["has_motif"] = df["gene_id"].isin(motif_genes)

    # 5. Plot MA plot
    plt.figure(figsize=(10, 6))
    plt.title(f"MA Plot: {os.path.basename(de_file)}")

    # Plot all genes
    plt.scatter(df["baseMean"], df["log2FoldChange"], c="gray", alpha=0.5, label="All genes")

    # Highlight motif-containing genes
    motif_subset = df[df["has_motif"]]
    plt.scatter(motif_subset["baseMean"], motif_subset["log2FoldChange"], 
                c="red", alpha=0.9, label="Genes with motif")
        
    # Axes settings
    plt.xscale("log")
    plt.xlabel("baseMean (log scale)")
    plt.ylabel("log2 Fold Change")
    plt.ylim(-10,10)
    plt.axhline(0, color="black", linestyle="--", linewidth=1)

    plt.legend()
    plt.tight_layout()


    
    # 6. Save plot
    out_plot = os.path.splitext(os.path.basename(de_file))[0] + "_MA_plot.png"
    plt.savefig(f"ma_plots/{out_plot}")
    plt.close()

    print(f"Saved MA plot for {de_file} as ma_plots/{out_plot}")

pval_threshold = 0.05  # You can adjust this
for de_file in de_files:
    # 3. Load DE result
    df = pd.read_csv(de_file)
    
    # Assume first column is gene name (some tools like DESeq2 export it that way)
    if df.columns[0] == "":
        df.rename(columns={df.columns[0]: "gene_id"}, inplace=True)
    else:
        df.rename(columns={df.columns[0]: "gene_id"}, inplace=True)

   # 1. Filter to only significant genes
    sig_df = df[df["pvalue"] < pval_threshold]
    
    # 2. Identify motif-positive genes within significant genes
    sig_df["has_motif"] = sig_df["gene_id"].isin(motif_genes)
    sig_motif = sig_df[sig_df["has_motif"]]
    sig_nonmotif = sig_df[~sig_df["has_motif"]]

    # 3. Plot
    plt.figure(figsize=(10, 6))
    plt.title(f"MA Plot (p < {pval_threshold}): {os.path.basename(de_file)}")

    # Plot significant genes without motif
    plt.scatter(sig_nonmotif["baseMean"], sig_nonmotif["log2FoldChange"], 
                c="gray", alpha=0.5, label="Significant (no motif)")

    # Plot significant genes with motif
    plt.scatter(sig_motif["baseMean"], sig_motif["log2FoldChange"], 
                c="red", alpha=0.9, label="Significant + motif")

    plt.xscale("log")
    plt.xlabel("baseMean (log scale)")
    plt.ylabel("log2 Fold Change")
    plt.axhline(0, color="black", linestyle="--", linewidth=1)
    plt.legend()
    plt.tight_layout()
    plt.ylim(-10,10)
    # Save plot
    out_plot_clean = os.path.splitext(os.path.basename(de_file))[0] + f"_MA_clean_p{pval_threshold}.png"
    plt.savefig(f"ma_plots/{out_plot_clean}")
    plt.close()

    print(f"Saved clean MA plot to ma_plots/{out_plot_clean}")






```
